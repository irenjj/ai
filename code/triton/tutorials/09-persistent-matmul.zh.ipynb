# 持久化矩阵乘法

本教程展示使用 Triton 实现持久化矩阵乘法内核，包括朴素实现、持久化实现、基于张量内存加速器（TMA）的实现，以及结合簇启动控制（CLC）与 TMA 的实现。

CLC 版本要求 NVIDIA GPU 的计算能力不低于 10.0。内核支持 FP16 和 FP8，但 FP8 实现仅适用于计算能力不低于 9.0 的 CUDA 设备。

教程在不同配置下对 Triton 和 cuBLAS 进行基准测试，并使用 Proton 性能分析器评估。可以通过命令行参数指定矩阵维度和迭代步长。
