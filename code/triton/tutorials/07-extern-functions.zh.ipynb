# Libdevice 外部函数

Triton 可以调用外部库中的自定义函数。本教程使用 libdevice 库，对张量计算反正弦。

有关可用函数的语义，请参考 [CUDA libdevice 用户指南](https://docs.nvidia.com/cuda/libdevice-users-guide/index.html) 或 [HIP 设备库源码](https://github.com/ROCm/llvm-project/tree/amd-staging/amd/device-libs/ocml/src)。

在 `libdevice.py` 中，执行相同运算但数据类型不同的函数会被组织在一起。例如，`__nv_asin` 和 `__nv_asinf` 都计算反正弦的主值，前者处理双精度浮点数，后者处理单精度浮点数。

Triton 会根据输入和输出类型，自动选择正确的底层设备函数。
